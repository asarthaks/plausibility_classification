```python
sent1 = "I've been waiting for a HuggingFace course for my whole life!."
sent2 = "I hate this!"
```

```python
sent1_ids = tokenizer.convert_tokens_to_ids(tokenizer.tokenize(sent1))
sent2_ids = tokenizer.convert_tokens_to_ids(tokenizer.tokenize(sent2))
```

```python
print(sent1_ids)
print(sent2_ids)
```

```python
sent2_len = len(sent2_ids)
new_sent2_ids = sent2_ids.copy()
for _ in range(len(sent1_ids) - len(sent2_ids)):
    new_sent2_ids.append(tokenizer.pad_token_id)
```

```python
print(sent1_ids)
print(new_sent2_ids)
```

```python
batches_ids = [sent1_ids, new_sent2_ids]
batches_ids
```

```python
attention_mask = list()
for batch_id in batches_ids:
    attention_mask.append([0 if item==tokenizer.pad_token_id else 1 for item in batch_id])
```
